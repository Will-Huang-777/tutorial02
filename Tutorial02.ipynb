```python
n, p, m_cols = 4, 3, 5

A = np.random.rand(n, p)
B = np.random.rand(p, m_cols)

C = np.dot(A, B)
print("A 的形状:", A.shape)
print("B 的形状:", B.shape)
print("结果形状:", C.shape)
print(C)
```
